## Cell 0 — Cài thư viện (chạy 1 lần)

```python
# Chạy 1 lần để cài các thư viện cần thiết
import subprocess, sys
pkgs = ['pandas', 'numpy', 'matplotlib', 'seaborn',
        'scikit-learn', 'xgboost', 'lightgbm', 'imbalanced-learn', 'tqdm']
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '--quiet'] + pkgs)
print('✓ Đã cài xong')
```

## Cell 1 — Load dữ liệu & tổng quan

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('data/features.csv')

print(f'Shape: {df.shape}')
print(f'\nPhân phối nhãn:')
print(df['label'].value_counts().rename({0: 'Legitimate (0)', 1: 'Phishing (1)'}))
print(f'\nSố đặc trưng: {df.shape[1] - 1}')
print(f'NaN tổng: {df.isnull().sum().sum()}')
df.describe().round(3)
```

## Cell 2 — Phân phối nhãn & Correlation Heatmap

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Biểu đồ phân phối nhãn
counts = df['label'].value_counts()
axes[0].bar(['Legitimate (0)', 'Phishing (1)'], counts.values,
            color=['#2196F3', '#F44336'], edgecolor='black')
axes[0].set_title('Phân phối nhãn (trước SMOTE)', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Số lượng URL')
for i, v in enumerate(counts.values):
    axes[0].text(i, v + 30, str(v), ha='center', fontweight='bold')

# Correlation heatmap (top 15 features by variance)
top_feats = df.drop(columns=['label']).var().nlargest(15).index.tolist() + ['label']
corr = df[top_feats].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, ax=axes[1], cmap='RdBu_r', center=0,
            annot=True, fmt='.2f', annot_kws={'size': 7},
            linewidths=0.5)
axes[1].set_title('Correlation Matrix (Top 15 Features)', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.savefig('data/eda_overview.png', dpi=150, bbox_inches='tight')
plt.show()
print('✓ Đã lưu: data/eda_overview.png')
```

## Cell 3 — Tiền xử lý: SMOTE + Train/Test Split

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE

# Tách feature và label
X = df.drop(columns=['label']).fillna(0).values
y = df['label'].values
feature_names = df.drop(columns=['label']).columns.tolist()

# Train/test split TRƯỚC khi SMOTE (quan trọng — không để dữ liệu synthetic vào test set)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# SMOTE chỉ áp dụng trên tập TRAINING
smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print(f'Train trước SMOTE: {X_train.shape[0]:,} ({y_train.sum()} phishing)')
print(f'Train sau  SMOTE:  {X_train_res.shape[0]:,} ({y_train_res.sum()} phishing)')
print(f'Test  (gốc):       {X_test.shape[0]:,} ({y_test.sum()} phishing)')

# Chuẩn hoá cho LR và SVM
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train_res)
X_test_sc  = scaler.transform(X_test)

print('\n✓ Tiền xử lý hoàn tất')
```

## Cell 4 — Logistic Regression

```python
import time
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, confusion_matrix

def evaluate(name, model, X_tr, X_te, y_tr, y_te, label=''):
    """Train model và in kết quả."""
    t0 = time.time()
    model.fit(X_tr, y_tr)
    train_time = time.time() - t0

    t0 = time.time()
    y_pred = model.predict(X_te)
    infer_ms = (time.time() - t0) / len(X_te) * 1000

    acc  = accuracy_score(y_te, y_pred)
    f1   = f1_score(y_te, y_pred)
    auc  = roc_auc_score(y_te, model.predict_proba(X_te)[:,1]) \
           if hasattr(model, 'predict_proba') else 0
    cm   = confusion_matrix(y_te, y_pred)
    tn, fp, fn, tp = cm.ravel()
    fpr  = fp / (fp + tn) if (fp + tn) > 0 else 0
    fnr  = fn / (fn + tp) if (fn + tp) > 0 else 0

    print(f'\n{'─'*50}')
    print(f'  {name} {label}')
    print(f'{'─'*50}')
    print(f'  Accuracy  : {acc:.4f}  ({acc*100:.2f}%)')
    print(f'  F1-Score  : {f1:.4f}')
    print(f'  ROC-AUC   : {auc:.4f}')
    print(f'  FPR       : {fpr:.4f}  ({fpr*100:.2f}%)')
    print(f'  FNR       : {fnr:.4f}  ({fnr*100:.2f}%)')
    print(f'  Train time: {train_time:.2f}s')
    print(f'  Infer/URL : {infer_ms:.3f} ms')

    return {
        'Model': name, 'Accuracy': round(acc, 4), 'F1': round(f1, 4),
        'ROC_AUC': round(auc, 4), 'FPR': round(fpr, 4), 'FNR': round(fnr, 4),
        'Latency_ms': round(infer_ms, 3)
    }

results = []

lr = LogisticRegression(max_iter=1000, random_state=42, n_jobs=-1)
results.append(evaluate('Logistic Regression', lr, X_train_sc, X_test_sc, y_train_res, y_test))
```

## Cell 5 — Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
results.append(evaluate('Random Forest', rf, X_train_res, X_test, y_train_res, y_test))
```

## Cell 6 — XGBoost ⭐ (Mô Hình Chính)

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    use_label_encoder=False,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)
results.append(evaluate('XGBoost', xgb, X_train_res, X_test, y_train_res, y_test,
                        label='⭐'))
```

## Cell 7 — Bảng So Sánh Kết Quả

```python
results_df = pd.DataFrame(results)

# Highlight giá trị tốt nhất mỗi cột
print('\n' + '='*65)
print('  BẢNG SO SÁNH KẾT QUẢ CÁC MÔ HÌNH')
print('='*65)
print(results_df.to_string(index=False))
print('='*65)

best = results_df.loc[results_df['F1'].idxmax()]
print(f'\n🏆 Mô hình tốt nhất theo F1: {best["Model"]} (F1={best["F1"]:.4f})')

# Lưu bảng kết quả
results_df.to_csv('data/model_comparison.csv', index=False)
print('✓ Đã lưu: data/model_comparison.csv')
```

## Cell 8 — Feature Importance (XGBoost)

```python
import matplotlib.pyplot as plt

importances = pd.Series(xgb.feature_importances_, index=feature_names)
top15 = importances.nlargest(15).sort_values()

fig, ax = plt.subplots(figsize=(8, 6))
top15.plot(kind='barh', ax=ax, color='#1976D2', edgecolor='black')
ax.set_title('Top 15 Feature Importance — XGBoost', fontsize=13, fontweight='bold')
ax.set_xlabel('Importance Score')
plt.tight_layout()
plt.savefig('data/feature_importance.png', dpi=150, bbox_inches='tight')
plt.show()
print('✓ Đã lưu: data/feature_importance.png')
```

## Cell 9 — Export Model (JSON cho Chrome Extension)

```python
import json, os

model_path = 'data/xgboost_model.json'
xgb.save_model(model_path)

size_kb = os.path.getsize(model_path) / 1024
print(f'✓ Model đã export: {model_path}')
print(f'  Kích thước: {size_kb:.1f} KB')
print(f'  (Yêu cầu DECUONG: <1 MB cho Chrome Extension)')

# Lưu danh sách tên đặc trưng (cần khi dùng trong extension)
with open('data/feature_names.json', 'w') as f:
    json.dump(feature_names, f, indent=2)
print(f'✓ Feature names: data/feature_names.json ({len(feature_names)} features)')
```
